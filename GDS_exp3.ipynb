### Data Generation for Traffic Congestion

```python
import folium
import pydeck as pdk
import pandas as pd
import numpy as np

# Generate synthetic data for traffic congestion in a city
def generate_traffic_data(num_points=100):
    np.random.seed(42)
    data = {
        'lat': np.random.uniform(34.0, 34.1, num_points), # Latitude around a city (e.g., Los Angeles)
        'lon': np.random.uniform(-118.3, -118.2, num_points), # Longitude around a city
        'traffic_intensity': np.random.randint(1, 10, num_points), # 1 (low) to 10 (high)
        'timestamp': pd.to_datetime('2024-07-20 08:00:00') + pd.to_timedelta(np.random.randint(0, 24 * 60, num_points), unit='min')
    }
    df = pd.DataFrame(data)
    return df

traffic_df = generate_traffic_data(num_points=500)
print("Generated Traffic Data (first 5 rows):")
print(traffic_df.head())
```

### 2D Visualization with Leaflet (via Folium)

```python
# Create a base map centered around the average latitude and longitude
center_lat = traffic_df['lat'].mean()
center_lon = traffic_df['lon'].mean()

m = folium.Map(location=[center_lat, center_lon], zoom_start=13)

# Add markers for each traffic point
for idx, row in traffic_df.iterrows():
    color = 'green' if row['traffic_intensity'] < 4 else \
            'orange' if row['traffic_intensity'] < 7 else 'red'

    folium.CircleMarker(
        location=[row['lat'], row['lon']],
        radius=5 + row['traffic_intensity'] * 0.5, # Radius based on intensity
        popup=f"Traffic Intensity: {row['traffic_intensity']}",
        color=color,
        fill=True,
        fill_color=color
    ).add_to(m)

# Display the map
m
```

### 3D Visualization with Deck.gl (via Pydeck)

```python
# Create a PyDeck Layer for the traffic data
# We'll use a ColumnLayer to represent traffic intensity as 3D columns
column_layer = pdk.Layer(
    "ColumnLayer",
    data=traffic_df,
    get_position=["lon", "lat"],
    get_elevation="traffic_intensity", # Height of the column based on intensity
    elevation_scale=50, # Scale the height for better visualization
    radius=50, # Radius of the columns
    get_fill_color=[
        "traffic_intensity * 25",
        "255 - traffic_intensity * 25",
        "100",
        180 # Transparency
    ], # Color based on intensity (e.g., green to red)
    pickable=True,
    auto_highlight=True
)

# Set the viewport location
view_state = pdk.ViewState(
    latitude=traffic_df['lat'].mean(),
    longitude=traffic_df['lon'].mean(),
    zoom=11,
    pitch=45, # Tilt the map for 3D effect
    bearing=0
)

# Create the PyDeck map
r = pdk.Deck(
    layers=[column_layer],
    initial_view_state=view_state,
    map_style='mapbox://styles/mapbox/light-v9', # Using a light map style
    tooltip={
        "text": "Traffic Intensity: {traffic_intensity}"
    }
)

# Display the PyDeck map
r
```

### Interactive Visualization with Plotly

```python
# Install Plotly if not already installed
!pip install plotly

import plotly.express as px

# Create an interactive scatter mapbox plot with Plotly
fig = px.scatter_mapbox(
    traffic_df,
    lat="lat",
    lon="lon",
    color="traffic_intensity",
    size="traffic_intensity",
    hover_name="timestamp",
    color_continuous_scale=px.colors.sequential.Inferno,
    zoom=11,
    title="Traffic Congestion Visualization with Plotly"
)

fig.update_layout(mapbox_style="open-street-map")
fig.update_layout(margin={"r":0,"t":50,"l":0,"b":0})

# Display the Plotly figure
fig.show()
```

### Large-Scale Geospatial Visualization with Kepler.gl

```python
# Install keplergl if not already installed
!pip install keplergl

from keplergl import KeplerGl

# Create a KeplerGl map object
map_kepler = KeplerGl(height=500)

# Add the traffic data to the map
# Kepler.gl automatically infers the data types and suggests layers
map_kepler.add_data(data=traffic_df, name='Traffic Data')

# Display the KeplerGl map
map_kepler
```

### 3D Globe Visualization with Cesium.js

Integrating Cesium.js directly within a Colab notebook for live interactive display is more complex as it requires a full web environment. However, we can generate an HTML file that uses Cesium.js to visualize the traffic data. You can then download and open this HTML file in your browser to see the 3D globe visualization.

```python
from IPython.display import HTML
import pandas as pd
import numpy as np
import json # Import json module for data serialization

# --- Re-defining traffic_df to ensure it's available ---
# This section is repeated to make sure traffic_df is defined
# in case previous cells were not run or the kernel was reset.
def generate_traffic_data(num_points=100):
    np.random.seed(42)
    data = {
        'lat': np.random.uniform(34.0, 34.1, num_points), # Latitude around a city (e.g., Los Angeles)
        'lon': np.random.uniform(-118.3, -118.2, num_points), # Longitude around a city
        'traffic_intensity': np.random.randint(1, 10, num_points), # 1 (low) to 10 (high)
        'timestamp': pd.to_datetime('2024-07-20 08:00:00') + pd.to_timedelta(np.random.randint(0, 24 * 60, num_points), unit='min')
    }
    df = pd.DataFrame(data)
    return df

traffic_df = generate_traffic_data(num_points=500)
# --------------------------------------------------------

# Prepare data for Cesium.js (simple point visualization)
cesium_points = []
for idx, row in traffic_df.iterrows():
    color = 'green' if row['traffic_intensity'] < 4 else \
            'orange' if row['traffic_intensity'] < 7 else 'red'

    # Cesium needs color in RGBA format (0-255 for R,G,B; 0-1 for A)
    rgba_color = [0, 0, 0, 0.8] # Default to semi-transparent black
    if color == 'green':
        rgba_color = [0, 255, 0, 0.8]
    elif color == 'orange':
        rgba_color = [255, 165, 0, 0.8]
    elif color == 'red':
        rgba_color = [255, 0, 0, 0.8]

    cesium_points.append({
        'lon': row['lon'],
        'lat': row['lat'],
        'height': row['traffic_intensity'] * 1000, # Scale intensity to height in meters
        'color': rgba_color,
        'popup': f"Traffic Intensity: {row['traffic_intensity']} "
    })

# HTML template for Cesium.js visualization
html_content = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="utf-8">
    <meta http-equiv="X-UA-Compatible" content="IE=edge">
    <meta name="viewport" content="width=device-width, initial-scale=1, -webkit-touch-callout=none, -webkit-user-select=none, -khtml-user-select=none, -moz-user-select=none, -ms-user-select=none, user-select:none;">
    <title>Cesium Traffic Visualization</title>
    <script src="https://cesium.com/downloads/cesiumjs/releases/1.118/Build/Cesium/Cesium.js"></script>
    <link href="https://cesium.com/downloads/cesiumjs/releases/1.118/Build/Cesium/Widgets/widgets.css" rel="stylesheet">
    <style>
        html, body, #cesiumContainer {{
            width: 100%;
            height: 100%;
            margin: 0;
            padding: 0;
            overflow: hidden;
        }}
    </style>
</head>
<body>
    <div id="cesiumContainer"></div>
    <script>
        Cesium.Ion.defaultAccessToken = 'YOUR_CESIUM_ION_ACCESS_TOKEN'; // Replace with your token

        const viewer = new Cesium.Viewer('cesiumContainer', {{
            // terrainProvider: Cesium.createWorldTerrain()
        }});

        const trafficData = {traffic_points_json};

        trafficData.forEach(point => {{
            viewer.entities.add({{
                position: Cesium.Cartesian3.fromDegrees(point.lon, point.lat, point.height),
                point: {{
                    pixelSize: 10,
                    color: Cesium.Color.fromBytes(point.color[0], point.color[1], point.color[2], point.color[3] * 255),
                    outlineColor: Cesium.Color.BLACK,
                    outlineWidth: 2
                }},
                label: {{
                    text: point.popup,
                    font: '14pt monospace',
                    style: Cesium.LabelStyle.FILL_AND_OUTLINE,
                    outlineWidth: 2,
                    verticalOrigin: Cesium.VerticalOrigin.BOTTOM,
                    pixelOffset: new Cesium.Cartesian2(0, -9)
                }}
            }}); // Escaped viewer.entities.add closing brace
        }});

        // Fly to the location
        viewer.camera.flyTo({{
            destination: Cesium.Cartesian3.fromDegrees(
                {avg_lon},
                {avg_lat},
                15000 // altitude in meters
            ),
            orientation: {{
                heading: Cesium.Math.toRadians(0.0),
                pitch: Cesium.Math.toRadians(-45.0),
                roll: 0.0
            }}
        }}); // Escaped viewer.camera.flyTo closing brace
    </script>
</body>
</html>
""".format(
    traffic_points_json=json.dumps(cesium_points),
    avg_lon=traffic_df['lon'].mean(),
    avg_lat=traffic_df['lat'].mean()
)

# Save the HTML content to a file
file_path = 'cesium_traffic_visualization.html'
with open(file_path, 'w') as f:
    f.write(html_content)

print(f"Cesium.js visualization saved to {file_path}")
print("Please replace 'YOUR_CESIUM_ION_ACCESS_TOKEN' with your actual Cesium Ion access token for full functionality.")

# Display a link to download the file (optional, for convenience in Colab)
HTML(f'<p>Download your Cesium visualization: <a href="{file_path}" download>cesium_traffic_visualization.html</a></p>')
```
